# 09 - Local forecast dashboard

Build an offline, standalone HTML dashboard with shared product-family, SKU, region, sales-channel, and customer filters. It includes executive, data-health, demand-explorer, anomaly-monitor, and forecast/accuracy views; historical-to-forecast units and estimated revenue; lifecycle/event indicators; and WAPE/bias slices. Channel/customer filters scope forecast SKU-region combinations by historical association, not customer-level allocation. No external runtime service or JavaScript package is required.

```python
from pathlib import Path
import sys

def find_project_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / "outputs" / "business_checks").is_dir():
            return candidate
    raise FileNotFoundError("Run notebooks 07 and 08 before building the dashboard.")

PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print(f"Project root: {PROJECT_ROOT}")
```

```python
from src.dashboard import build_local_dashboard

dashboard = build_local_dashboard(PROJECT_ROOT)
dashboard
```

```python
dashboard_path = Path(dashboard["dashboard_path"])
html = dashboard_path.read_text(encoding="utf-8")

assert dashboard["forecast_rows_embedded"] == 58_219
assert dashboard["forecast_months"] == 19
assert dashboard["external_dependencies"] == 0
assert "__DASHBOARD_DATA__" not in html
assert "family-filter" in html and "region-filter" in html and "sku-filter" in html
assert "<script src=" not in html.lower()
assert "<link rel=\"stylesheet\" href=\"http" not in html.lower()
print(f"Dashboard ready: {dashboard_path} ({dashboard['bytes']:,} bytes)")
```

Open `outputs/dashboard/index.html` directly in a browser. The forecast rows are embedded in the page, so it remains usable offline and does not require serving CSV files.
